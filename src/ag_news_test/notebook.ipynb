```python
from pathlib import Path

from datasets import load_dataset
from src.folder_organizer.classifier import AIFileClassifier, ClassesDefinition, AppConfig
from src.folder_organizer.prompts import DirectPromptStrategy
import json
```

```python
classes_definition = ClassesDefinition.from_yaml(Path("classes.yaml"))
app_config = AppConfig.from_yaml(Path("app_config.yaml"))
classifier = AIFileClassifier(classes_definition=classes_definition, app_config=app_config,
                              prompt_strategy=DirectPromptStrategy())
```

```python
dataset = load_dataset("sh0416/ag_news", split="test[:10]")
```

```python
label_mapping = {1: "World", 2: "Sports", 3: "Business", 4: "Sci/Tech"}
```

```python
classifications = []
for item in dataset:
    title = item["title"]
    text = f"{item['title']}: {item['description']}"
    label = item["label"]
    prediction = await classifier.classify(filename=title, content=text)
    classifications.append((label_mapping[label], prediction))
```

```python

for label, prediction in classifications:
    print(f"True Label: {label}\nPrecited label: {prediction.predicted_class}\nPredicted Result:\n {json.dumps(prediction.to_dict(), sort_keys=True, indent=2)}\n{'-'*50}")
```
